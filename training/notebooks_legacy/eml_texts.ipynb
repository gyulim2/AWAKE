```python
import email
from email import policy
from email.parser import BytesParser
from bs4 import BeautifulSoup  # HTML 본문 처리를 위한 라이브러리

# .eml 파일 경로 설정
eml_file_path = r"C:\Users\brady\Downloads\정상이메일testdata2.eml"

def extract_email_body_recursive(part):
    """재귀적으로 파트를 탐색하여 본문 추출"""
    if part.is_multipart():
        # 각 하위 파트를 순회
        for subpart in part.iter_parts():
            extract_email_body_recursive(subpart)  # 하위 파트를 재귀적으로 탐색
    else:
        # 텍스트 본문을 추출하는 부분
        if part.get_content_type() == 'text/plain':
            body = part.get_payload(decode(True)).decode(part.get_content_charset(), errors="replace")
            print(f"텍스트 본문:\n{body}")
        elif part.get_content_type() == 'text/html':
            html_content = part.get_payload(decode=True).decode(part.get_content_charset(), errors="replace")
            soup = BeautifulSoup(html_content, 'html.parser')
            body = soup.get_text()
            print(f"HTML 본문 (텍스트로 변환):\n{body}")
        else:
            print(f"텍스트가 아닌 파트입니다. MIME 타입: {part.get_content_type()}")

def extract_email_body(eml_file_path):
    """이메일 파일에서 본문 추출"""
    with open(eml_file_path, 'rb') as f:
        msg = BytesParser(policy=policy.default).parse(f)

    # 이메일 제목 추출
    subject = msg['subject']
    print(f"이메일 제목: {subject}\n")

    if msg.is_multipart():
        print("멀티파트 이메일이 감지되었습니다. 파트를 순회 중입니다.")
        extract_email_body_recursive(msg)  # 재귀적으로 모든 파트를 탐색
    else:
        # 멀티파트가 아닌 경우
        body = msg.get_payload(decode=True).decode(msg.get_content_charset(), errors="replace")
        print(f"단일 파트 본문:\n{body}")

# 실행
extract_email_body(eml_file_path)
```

```python
import pdfplumber  # PDF 파일 처리
import docx  # DOCX 파일 처리

# .eml 파일 경로 설정
eml_file_path = r"C:\Users\brady\Downloads\정상이메일testdata2.eml"

def extract_attachments(eml_file_path):
    """이메일 파일에서 첨부파일을 추출하고 텍스트를 추출"""
    with open(eml_file_path, 'rb') as f:
        msg = BytesParser(policy=policy.default).parse(f)

    if msg.is_multipart():
        print("첨부파일을 찾는 중입니다...")
        for part in msg.iter_parts():
            if part.get_content_disposition() and 'attachment' in part.get_content_disposition():
                filename = part.get_filename()
                if filename:
                    print(f"\n첨부파일 발견: {filename}")
                    attachment_data = part.get_payload(decode=True)
                    # 파일 확장자를 소문자로 변환한 후 처리
                    if filename.lower().endswith(".pdf"):
                        print(f"첨부파일 제목: {filename}")
                        extract_pdf_text(attachment_data)
                    elif filename.lower().endswith(".docx"):
                        print(f"첨부파일 제목: {filename}")
                        extract_docx_text(attachment_data)
                    else:
                        print(f"지원되지 않는 파일 형식입니다: {filename}")
    else:
        print("첨부파일이 없습니다.")

def extract_pdf_text(pdf_data):
    """PDF 파일에서 텍스트 추출"""
    try:
        with pdfplumber.open(BytesIO(pdf_data)) as pdf:
            text = ''
            for page in pdf.pages:
                page_text = page.extract_text()
                if page_text:
                    text += page_text + "\n"
            if text:
                print(f"PDF 파일에서 추출한 텍스트:\n{text}")
            else:
                print("PDF 파일에서 텍스트를 추출하지 못했습니다. 빈 파일이거나 이미지로만 구성된 PDF일 수 있습니다.")
    except Exception as e:
        print(f"PDF 파일 처리 중 오류 발생: {e}")

def extract_docx_text(docx_data):
    """DOCX 파일에서 텍스트 추출"""
    try:
        doc = docx.Document(BytesIO(docx_data))
        text = '\n'.join([para.text for para in doc.paragraphs])
        if text:
            print(f"DOCX 파일에서 추출한 텍스트:\n{text}")
        else:
            print("DOCX 파일에서 텍스트를 추출하지 못했습니다.")
    except Exception as e:
        print(f"DOCX 파일 처리 중 오류 발생: {e}")

# 첨부파일 추출 함수 실행
extract_attachments(eml_file_path)
```
